```python
# AI Secretary Agent for Multiple Appointments
# Author: Angelo Sorte
# Description: This script extracts multiple appointment requests from text messages and generates an .ics calendar file.

# Install required packages
!pip install spacy==3.7.2 dateparser ics
!python -m spacy download en_core_web_sm

import spacy
from dateparser.search import search_dates
from ics import Calendar, Event
from datetime import datetime, time

# Load English NLP model for Named Entity Recognition
nlp = spacy.load("en_core_web_sm")

# Example input message (this simulates an email or WhatsApp message)
message = "Hi, can we schedule meetings next Tuesday at 3 pm and Wednesday morning at 10?"

# Step 1: Process the message with spaCy
# This identifies entities such as dates, times, people, organizations, etc.
doc = nlp(message)

# Step 2: Extract all date/time expressions from the text
results = search_dates(message)

if results:
    print("All detected dates:")
    for match, dt in results:
        print(f"- '{match}' parsed as {dt}")
else:
    print("No valid dates found in the message.")

# Step 3: Create a new calendar with all found events
if results:
    c = Calendar()
    for idx, (match, dt) in enumerate(results, start=1):
        e = Event()
        e.name = f"Meeting {idx}"

        # If time is missing, set default to 09:00
        if dt.time() == time(0, 0):
            dt = datetime.combine(dt.date(), time(9, 0))

        e.begin = dt
        e.duration = {"minutes": 60}  # Default duration: 1 hour
        c.events.add(e)

    # Save the events to an .ics file
    with open("appointments.ics", "w") as f:
        f.writelines(c)
    print("ICS calendar file created: appointments.ics")

# Step 4: Display recognized entities for learning purpose
print("\nNamed Entities detected by spaCy:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
